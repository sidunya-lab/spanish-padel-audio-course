```
# === ШАГ 1: СБОРКА И ПРОСЛУШИВАНИЕ ТЕСТОВОЙ ГЛАВЫ (БЕЗ АРХИВАЦИИ) ===
import os, glob, zipfile, shutil
from google.colab import drive
from IPython.display import Audio, display

GEMINI_API_KEY = ''  # <-- ВСТАВЬ СВОЙ КЛЮЧ GEMINI API СЮДА В КАВЫЧКИ
os.environ['GEMINI_API_KEY'] = GEMINI_API_KEY.strip()

drive.mount('/content/drive')
!pip install -q google-genai edge-tts faster-whisper

folder_matches = glob.glob('/content/drive/MyDrive/*Испанский для падела — v4*')
assert folder_matches, 'Не найдена папка Испанский для падела — v4 на Google Диске'
gdrive_folder = folder_matches[0]
zips = glob.glob(os.path.join(gdrive_folder, '*.zip'))
os.makedirs('/content/padel_v4', exist_ok=True)
with zipfile.ZipFile(sorted(zips)[-1], 'r') as zf:
    zf.extractall('/content/padel_v4')

!python3 /content/padel_v4/build_course.py --project-dir /content/padel_v4 --only-test --out-dir /content/dist_test
mp3_files = glob.glob('/content/dist_test/chapters/*.mp3')
shutil.copy2(mp3_files[0], os.path.join(gdrive_folder, os.path.basename(mp3_files[0])))
print('Готовый MP3 сохранён в Google Диск без архивации:', os.path.basename(mp3_files[0]))
display(Audio(mp3_files[0]))
```

```
# === ШАГ 2: ПОЛНАЯ СБОРКА ВСЕХ 27 ГЛАВ + РЕЖИМ ТЕНЬ В GOOGLE ДИСК ===
!python3 /content/padel_v4/build_course.py --project-dir /content/padel_v4 --out-dir /content/dist_full --with-shadowing
out_target = os.path.join(gdrive_folder, 'Готовое_аудио_MP3_v4')
if os.path.exists(out_target):
    shutil.rmtree(out_target)
shutil.copytree('/content/dist_full', out_target)
print('ВЕСЬ КУРС СОБРАН И СОХРАНЁН НА GOOGLE ДИСКЕ:', out_target)
```
